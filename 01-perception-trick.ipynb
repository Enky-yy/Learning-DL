```python
from sklearn.datasets import  make_classification
import numpy as np
X,y = make_classification(n_samples=80, n_features=2, n_informative=1, n_redundant=0, n_classes=2 , n_clusters_per_class=1, random_state=39, hypercube=False, class_sep=10)
```

```python
import  matplotlib.pyplot as plt
```

```python
plt.Figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y, cmap="winter", s=100)
```

```python
def steps(z):
    return 1 if z>0 else 0
```

```python
def perception(X,y):
    X= np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr=0.1

    for i in range(1000):
        j =np.random.randint(0,79)
        y_hat = steps(np.dot(X[j],weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]

    return  weights[0], weights[1:]
```

```python
intercept_, coef_ = perception(X,y)
```
