```python
df= pd.read_csv("/kaggle/input/credit-card-customer-churn-prediction/Churn_Modelling.csv")
```

```python
df.head(10)
##%%
df.shape
##%%
df.info()
##%%
df.duplicated().sum()
##%%
df.Exited.value_counts()
##%%
df.Geography.value_counts()
##%%
df.Gender.value_counts()
##%%
df.drop(columns=['RowNumber',"CustomerId","Surname"], inplace=True)
##%%
df.head()
##%%
df = pd.get_dummies(df,columns=['Geography',"Gender"],drop_first=True)
##%%
```

```python
from   sklearn.model_selection import train_test_split
##%%
(train_df , test_df)= train_test_split(df,test_size= 0.2 ,random_state=1)
##%%
df.columns
##%%
inputs = ['CreditScore', 'Age', 'Tenure', 'Balance', 'NumOfProducts', 'HasCrCard',
       'IsActiveMember', 'EstimatedSalary','Geography_Germany',
       'Geography_Spain', 'Gender_Male']
##%%
targets = ["Exited"]
##%%
train_df[inputs].shape
##%%
```

```python
from sklearn.preprocessing import StandardScaler
##%%
df
##%%
scaler = StandardScaler().fit(df[inputs])
##%%
train_df[inputs] = scaler.transform(train_df[inputs])
##%%
test_df[inputs] = scaler.transform(test_df[inputs])
##%%
train_df
##%%
```

```python
import tensorflow
from keras import Sequential
from keras.layers import Dense
##%%
model = Sequential()

model.add(Dense(3,activation='sigmoid',input_dim=11))
model.add(Dense(1,activation='sigmoid'))
##%%
model.summary()
##%%
model.compile(optimizer='Adam',loss='binary_crossentropy',metrics=['accuracy'])
##%%
model.fit(train_df[inputs],train_df[targets],batch_size=50,epochs=100,verbose=1,validation_split=0.2)
```

```python
preds = model.predict(test_df[inputs])
```

```python
preds = preds.argmax(axis=-1)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(test_df[targets],preds)
```
